1. The number of students participating in the discussion is described by Poisson distribution with unknown parameter λ, i.e. $P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$, where $x \geq 0$. Please use MLE to estimate the parameter λ by using the number of students in recent 8 discussions: 17, 11, 19, 14, 15, 21, 18, 13  (we assume that they are independent observations). Note: you need to write out the likelihood function and then find the MLE instead of applying the formula from the lecture or textbook.

In order to solve this problem, we need to determine the likelihood function for this Poisson distribution with 8 distinct k values for each input $k_1, ..., k_8$:

$L(λ|k_1, ..., k_8) = \prod_{i=1}^8 \frac{\lambda^{k_i} e^{-\lambda}}{k_i!}$

To simplify the differentiation needed, we can take the logarithm of the likelihood function to obtain the log-likelihood:

$\ln L(λ|k_1, ..., k_8) = \sum_{i=1}^8 (\ln(\frac{\lambda^{k_i}}{k_i!}) -\lambda) = \sum_{i=1}^8 (k_i\ln(\lambda) -\lambda - \ln(k_i!) )$

To find the maximum of the log-likelihood, we can set the first derivative of the above equation equal to 0 and find the corresponding $\lambda$ value, which is the maximum value of the function:

$\frac{d}{d\theta} \log L(\theta) = \sum_{i=1}^8 (\frac{k_i}{\lambda} -1) = 0 \quad \rightarrow \quad \frac{k_1+...+k_8}{\lambda} - 8= 0 \quad \rightarrow \quad λ = \frac{17 + 11 + 19 + 14 + 15 + 21 + 18 + 13}{8} = 16$


2. Suppose the grade of midterm1 of a class is normal with unknown 𝜇 and known variance $15^2$. Suppose that we know the prior $\mu$ ~ $N(85, 5^2)$. Given 10 student's grades: 61, 65, 76, 84, 87, 89, 90, 94, 94, 100, what is the MAP estimate of 𝜇?

In order to determine the MAP estimate, we need to determine the equation to solve for $P(μ|x)$ uses Bayes's rule:

$P(μ|x) = \frac{P(x|\mu) \cdot P(\mu)}{P(x)}$

Since we are solving for the value for $\mu$ to maximize the above equation, we can get rid of any values that don't contain μ. Therefore, we need to maximize $P(x|\mu) \cdot P(\mu)$.

The equation for $P(x|\mu)$ follows the pdf of a normal distribution:

$P(x|\mu) = \frac{1}{2 π σ} e^{\frac{-(x-μ)^2}{2σ}} = \frac{1}{30 π} e^{\frac{-(x-μ)^2}{30}} \quad → \quad P(x_1, ..., x_{10}|μ) = \prod_{i=1}^{10} \frac{1}{30 π} e^{\frac{-(x_i-μ)^2}{30}}$

The log of the above function is:

$\ln P(x_1, ..., x_{10}|\mu) = \sum_{i=1}^{10} (\frac{-(x_i-μ)^2}{30} + \ln(\frac{1}{30 π}))$

\\
The equation for $P(μ)$ follows the pdf of a normal distribution:

$P(\mu) = \frac{1}{2 π σ_0} e^{\frac{-(\mu-μ_0)^2}{2σ_0}} = \frac{1}{10 π} e^{\frac{-(\mu-85)^2}{10}}$

The log of the above function is:

$\ln P(\mu) = \frac{-(\mu-85)^2}{10} + \ln(\frac{1}{10 π})$

\\
Using the above, we can maximize the posterior expression $P(x|\mu) \cdot P(\mu)$ by taking the log of the expression, taking the first derivative, and setting it equal to 0:

$\ln(P(x_1, ..., x_{10}|\mu) \cdot P(\mu)) = \sum_{i=1}^{10} (\frac{-(x_i-μ)^2}{30} + \ln(\frac{1}{30 π})) + \frac{-(\mu-85)^2}{10} + \ln(\frac{1}{10 π})$
$\rightarrow \quad \frac{d}{d\mu} \ln(P(x|\mu) \cdot P(\mu)) = \sum_{i=1}^{10} \frac{x_i-μ}{15} + \frac{-(\mu-85)}{5} = 0$
$\quad \rightarrow \quad \hat{\mu} = \frac{840 + 255}{13} = 84.23$
